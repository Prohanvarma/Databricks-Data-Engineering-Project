## **Remove unused files using VACUUM**

VACUUM Command

Used to remove old, unused files from the Delta Lake to free up storage. Permanently deletes files that are no longer referenced in the transaction log and older than retention threshold (7 days)

**1. Check the table history**

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

2. Run VACUUM

```sql
VACUUM demo.delta_lake.optimize_stock_prices;
```

```sql
SET delta.retentionDurationCheck.enabled = false;
VACUUM demo.delta_lake.optimize_stock_prices RETAIN 0 HOURS;
```
